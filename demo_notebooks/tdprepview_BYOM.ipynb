# train model

```python
df_ADS_training = DF_ADS_training.to_pandas(all_rows=True)
```

```python
feature_cols = list(df_ADS_training.columns[3:])
```

```python
feature_cols
```

```python
X = df_ADS_training[feature_cols].values
y = df_ADS_training[target].values
```

```python
#!pip install catboost
```

Next, we train a supervised machine learning model. While many models could work, we choose CatBoost for its strong performance with minimal tuning required.

```python
import catboost
```

```python
model_catb = catboost.CatBoostClassifier(iterations=100)

model_catb.fit(X, y)
```

The next step is to convert the model to the ONNX format, making it compatible for deployment within Vantage. We then test the model by running it in a local ONNX runtime to ensure it works as expected. Finally, we upload the model into a table for further use.

```python
onnxfile_catb = "catb-model.onnx"
model_catb_id = "catb_model"
```

```python
my_export_params ={
        'onnx_domain': 'ai.catboost',
        'onnx_model_version': 1,
        'onnx_doc_string': 'model for churn',
        'onnx_graph_name': 'CatBoostModel_for_Churn'
    }
```

```python
model_catb.save_model(
    onnxfile_catb,
    format="onnx",
    export_parameters=my_export_params)
```

```python
model_table_name = "churn_models"
try:
    tdml.db_drop_table(model_table_name)
except:
    pass
```

```python
tdml.save_byom(model_id = model_catb_id, 
               model_file =  onnxfile_catb, 
               table_name =  model_table_name)
```

# test run of Preditions with Scoring ADS

```python
DF_ADS_scoring = tdml.DataFrame(tdml.in_schema(output_schema, view_ADS_scoring))
DF_ADS_scoring
```

```python
tdml.configure.byom_install_location = "mldb"
```

```python
onnxpred_obj = tdml.ONNXPredict(
    newdata = DF_ADS_scoring,
    modeldata = tdml.DataFrame(model_table_name),
    accumulate = [key],
    overwrite_cached_models = "*",
    model_input_fields_map = f"features={','.join(feature_cols)}" 
)
```

```python
onnxpred_obj.result
```

# Deploy Model Endpoint as View

```python
onnxpredict_query = onnxpred_obj.show_query()
print(onnxpredict_query)
```

```python
tdml.execute_sql(f"""
REPLACE VIEW churn_predictions AS
(
{onnxpredict_query}
)
""")
```
